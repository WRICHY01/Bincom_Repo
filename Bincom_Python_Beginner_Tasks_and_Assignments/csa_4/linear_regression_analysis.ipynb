```python
mean_hours = df["Hours"].mean()
mean_scores = df["Scores"].mean()


mean_hours, mean_scores
```

##### Using the formula from the attached image in the CSA_4 Folder, The slope and Intercept was gotten thus:

```python
x_dev = (df["Scores"] - mean_scores)
x_dev_squared = x_dev ** 2
y_dev = (df["Hours"] - mean_hours)

x_dev, x_dev_squared, y_dev
```

#### You can use the formula from the image in CSA_4 folder for reference.

#### where,
   * a = slope
   * i = intercept

```python
slope = sum(x_dev * y_dev)/sum(x_dev_squared)
intercept = mean_hours - slope * mean_scores

slope, intercept
```

```python
y_pred = intercept + slope * df["Scores"]

y_pred
```

```python
plt.scatter(df["Scores"], df["Hours"])
plt.plot(df["Scores"], y_pred, color="green", linewidth=3, label="regression_line")
plt.title(
    "Impact of Study Time on Student Performance\n(Linear Regression Analysis)",
    fontsize=14,
    fontweight="bold",
    pad=15,
)
plt.xlabel("Scores")
plt.ylabel("Hours")
plt.tight_layout()
plt.legend(loc="upper left", frameon=True, facecolor="white", edgecolor="none")

# Save the figure
plt.savefig(save_path + "/linear_regression_of_students_score_showing_positive_correlation_wrt_time_hours.png", dpi=300, bbox_inches="tight")

plt.show()
```
